## 2 Crear el chatbot utilizando TFIDF y similitud del coseno

```python
!pip install spacy --quiet
!python -m spacy download es_core_news_md --quiet
```

```python
import numpy as np
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from spacy.lang.es.stop_words import STOP_WORDS
```

```python
nlp = spacy.load("es_core_news_md")

# Se separa el dataset
preguntas  = df["pregunta"].tolist()
respuestas = df["respuesta"].tolist()
```

```python
CONSERVAR = {"no", "llegó", "sin", "más", "antes", "hasta"}
```

```python
# Funcion de preprocesamiento
def preprocesar(texto):
    doc = nlp(texto.lower())
    return " ".join(
        token.lemma_.lower()
        for token in doc
          if not token.is_punct
          and not token.is_space
        and (token.is_alpha or token.like_num)
        and (not token.is_stop or token.lower_ in CONSERVAR)
    )
```

```python
# Se preprocesan las preguntas
preguntas_limpias = []
for t in preguntas:
   p = preprocesar(t)
   preguntas_limpias.append(p)
```

```python
# Se crea la matriz corpus
vectorizer = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b")
matriz_corpus = vectorizer.fit_transform(preguntas_limpias)
```

```python
# Verificacion general
vocab = vectorizer.get_feature_names_out()
print(len(vocab), "términos en el vocabulario")
print(sorted(vocab)[:40])
print("Forma de la matriz:", matriz_corpus.shape)
```

```python
# Comparativa post conversion
for orig, limpia in zip(preguntas[:5], preguntas_limpias[:5]):
    print(f"ORIG:   {orig}")
    print(f"LIMPIA: {limpia}\n")
```

```python
MENSAJE_FALLBACK = "No tengo información sobre eso. ¿Podés reformular tu consulta?"

def responder(consulta, umbral):
    # limpiar la consulta
    consulta_limpia = preprocesar(consulta)

    # vectorizar
    vec = vectorizer.transform([consulta_limpia])

    # 3similitud coseno entre la consulta y las 28 preguntas
    sims = cosine_similarity(vec, matriz_corpus)[0]

    # índice de la pregunta más parecida y su similitud
    i = np.argmax(sims)
    mejor = sims[i]

    if mejor >= umbral:
        return respuestas[i], mejor, preguntas[i]
    else:
        return MENSAJE_FALLBACK, mejor, preguntas[i]
```

## 3. Chatbot - Embeddings

Elección del modelo: Se utilizó spaCy (es_core_news_md) porque ofrece vectores de palabras preentrenados en español (300 dimensiones) integrados en el mismo pipeline ya usado para el preprocesamiento del chatbot 1. Esto permite mantener una única dependencia y comparar ambos enfoques en igualdad de condiciones.


```python
#Se carga modelo en español
nlp = spacy.load("es_core_news_md")

# Se obtiene del dataset las preguntas y respuestas
preguntas  = df["pregunta"].tolist()
respuestas = df["respuesta"].tolist()
```

```python
# Se obtiene para cada pregunta el vector generado por el modelo
matriz = []
for t in preguntas:
  doc = nlp(t)
  vect = doc.vector
  matriz.append(vect)

# Se obtiene la matriz de embeddings
matriz = np.array(matriz)
```

```python
# Verificacion tamaño (cada fila un vector corresp a una pregunta)
matriz.shape
```

```python
MENSAJE_FALLBACK = "No tengo información sobre eso. ¿Podés reformular tu consulta?"

def responder_emb(consulta,umbral):

    # Se vectoriza mediante el modelo
    doc = nlp(consulta)

    vect_resp = doc.vector

    # Se ajusta vect
    vect_resp = vect_resp.reshape(1,-1)

    # Se calcula la similitud coseno
    sims = cosine_similarity(vect_resp, matriz)[0]

    # Se toma el indice la pregunta mas parecida y su similitd
    i = np.argmax(sims)
    mejor = sims[i]

    if mejor >= umbral:
        return respuestas[i], mejor, preguntas[i]
    else:
        return MENSAJE_FALLBACK, mejor, preguntas[i]
```

## 4. Prueba de Chatbots

```python
consultas_demo = [
    ("¿Cómo anulo una reserva?",                  "¿Puedo cancelar mi reserva?"),
    ("¿Me reintegran la plata si no puedo ir?",   "¿Me devuelven el dinero si cancelo?"),
    ("¿A qué hora puedo entrar al depto?",        "¿A qué hora es el check-in?"),
    ("¿Tienen alojamientos pet friendly?",        "¿Aceptan mascotas en los departamentos?"),
    ("¿Cuál es el valor por noche del Centro?",   "¿Cuánto sale el Depto Centro por noche?"),
    ("q onda si no llego al check in",            "¿Qué pasa si no me presento el día del check-in?"),
    ("hay internet?",                             "¿Los departamentos tienen wifi?"),
    ("tienen algo para dos personas q sea barato","¿Cuál es el alojamiento más económico?"),
    ("¿Cuál es la capital de Alemania?",           None),
    ("¿Qué temperatura hay hoy?",                None),
]
```

```python
# Se procesa cada pregunta y se compara la respuesta
for consulta, esperada in consultas_demo:
    resp1, sim1, preg1 = responder(consulta, 0.7) #TDIDF
    resp2, sim2, preg2 = responder_emb(consulta, 0.7) #Embedding

    c1 = resp1 != MENSAJE_FALLBACK
    ok1 = (not c1) if esperada is None else (c1 and preg1 == esperada)

    c2 = resp2 != MENSAJE_FALLBACK
    ok2 = (not c2) if esperada is None else (c2 and preg2 == esperada)

    tag1 = "Ok" if ok1 else "Error"
    tag2 = "Ok" if ok2 else "Error"

    print(f"CONSULTA: {consulta}")
    print(f"  TF-IDF     [{sim1:.2f}] ({tag1}): {resp1[:120]}")
    print(f"  Embeddings [{sim2:.2f}] ({tag2}): {resp2[:120]}")
    print()

```

```python
# Resumen de prueba
total = len(consultas_demo)

# Se inicializan contadores
ok_tfidf = 0
ok_emb = 0

for consulta, esperada in consultas_demo:
    resp1, sim1, preg1 = responder(consulta, 0.7)
    resp2, sim2, preg2 = responder_emb(consulta, 0.70)
    c1 = resp1 != MENSAJE_FALLBACK
    ok_tfidf += (not c1) if esperada is None else (c1 and preg1 == esperada)
    c2 = resp2 != MENSAJE_FALLBACK
    ok_emb += (not c2) if esperada is None else (c2 and preg2 == esperada)

print("=" * 60)
print(f"RESUMEN: {total} consultas")
print(f"  TF-IDF:     {ok_tfidf}/{total} correctas — {total - ok_tfidf}/{total} incorrectas")
print(f"  Embeddings: {ok_emb}/{total} correctas — {total - ok_emb}/{total} incorrectas")
```

### 4.1 Prueba interactiva

Permite probar ambos chatbots en vivo con cualquier consulta y comparar lo que devuelve cada uno junto a su similitud coseno.

```python
UMBRAL = 0.7

print("Probá el chatbot escribiendo tus consultas. Escribí 'salir' para terminar.\n")

while True:
    consulta = input("Vos > ").strip()
    if consulta.lower() in ("salir", "exit", "quit", ""):
        print("Chat terminado.")
        break

    r_tfidf, s_tfidf, p_tfidf = responder(consulta, UMBRAL)
    r_emb,   s_emb,   p_emb   = responder_emb(consulta, UMBRAL)

    print(f"\n>> TF-IDF     [sim={s_tfidf:.2f}] → match con: '{p_tfidf}'")
    print(f"   {r_tfidf}\n")
    print(f">> Embeddings [sim={s_emb:.2f}] → match con: '{p_emb}'")
    print(f"   {r_emb}")
    print("=" * 70 + "\n")
```

---


 Contrariamente a lo esperado, el chatbot basado en TF-IDF superó al de embeddings en este escenario. El resultado se explica por las características del corpus y la técnica de vectorización utilizada.



 TF-IDF aplica eliminación de stopwords por lo que al remover palabras funcionales  el vector solo contiene palabras con carga relevante. Esto permite que la similitud coseno refleje coincidencia real.

 Su principal debilidad es si la consulta no comparte tokens exactos con ninguna pregunta del dataset, la similitud cae a cero y el chatbot no puede responder.

----
los embeddings fallaron por que promedia los vectores de todas las palabras de la oración, incluyendo artículos, preposiciones y pronombres interrogativos.
 Se generan matches temáticamente cercanos pero incorrectos Los embeddings capturan que "precio" y "económico" están en el mismo campo semántico, pero no distinguen entre "¿Cuánto sale el Depto Centro?" y "¿Cuál es el más económico?".

---

#### Propuesta de mejora

 El problema central de los embeddings es el promedio indiscriminado de todos los vectores de palabras. Una mejora posible sería **filtrar stopwords antes de promediar**, conservando solo los vectores de palabras con contenido semántico.

#### Conclusión final

 Para un chatbot de recuperación de información con un corpus reducido y consultas en español, TF-IDF con preprocesamiento adecuado resultó más confiable que los embeddings promediados de spaCy.
